```python
import pandas as pd
import numpy as np
```

```python
df_pred = pd.read_csv('../data/processed/predicted_synthetic_compounds.csv')
```

```python
R_O = 128  # Ionic radius of O^2- in pm
tolerance_factor = ((df_pred['R-A1'] + df_pred['R-A2'])/2 + R_O) / (np.sqrt(2) * ((df_pred['R-B1'] + df_pred['R-B2'])/2 + R_O))
df_pred['tolerance_factor'] = tolerance_factor
octahedral_factor = ((df_pred['R-B1'] + df_pred['R-B2'])/2) /  R_O
df_pred['octahedral_factor'] = octahedral_factor
def calculate_new_tau_vectorized(df):
    rb = (df['R-B1'] + df['R-B2']) / 2
    ra = (df['R-A1'] + df['R-A2']) / 2
    rx = R_O
    na = df['A1_OS']

    invalid = (ra <= 0) | (rb <= 0) | (rx <= 0) | np.isclose(ra, rb)

    ratio = ra / rb
    with np.errstate(divide='ignore', invalid='ignore'):
        term1 = rx / rb
        term2 = na * (na - ratio / np.log(ratio))
        tau = term1 - term2


    tau[invalid] = np.nan
    return tau

df_pred['new_tau'] = calculate_new_tau_vectorized(df_pred)
```

```python
df_ABO3 = df_pred[(df_pred['type'] == 'ABO3') & (df_pred['stable'] == 1) & (df_pred['Insulator'] == 1)]
df_A2B1B2O6 = df_pred[(df_pred['type'] == 'A2B1B2O6') & (df_pred['stable'] == 1) & (df_pred['Insulator'] == 1)]
df_A1A2B2O6 = df_pred[(df_pred['type'] == 'A1A2B2O6') & (df_pred['stable'] == 1) & (df_pred['Insulator'] == 1)]
df_A1A2B1B2O6 = df_pred[(df_pred['type'] == 'A1A2B1B2O6') & (df_pred['stable'] == 1) & (df_pred['Insulator'] == 1)]
```

```python
nan_elements = []
nan_oxidations = []

def add_selected_ionic_radii(df, ion_dict, op_corr):
    """
    Add new columns S_IR-A1, S_IR-A2, S_IR-B1, S_IR-B2 to df.
    Radius chosen based on oxidation number and coordination number (corr):
        - Prefer corr == 12 if available
        - Otherwise use the highest corr value
    """
    
    def get_radius(element, ox_state):
        """Return ionic radius for given element and oxidation state."""
        if pd.isna(element) or pd.isna(ox_state):
            return None
        element = str(element)
        try:
            ox_state = int(ox_state)
        except ValueError:
            return None
        
        if element not in ion_dict:
            if element not in nan_elements:
                nan_elements.append(element)

            return None
            
        # Filter by oxidation number
        data = [d for d in ion_dict[element] if d["oxidation_num"] == ox_state]
        if not data:
            temp = (element, ox_state)
            if temp not in nan_oxidations:
                nan_oxidations.append(temp)
            return None
        
        # Prefer corr == 12
        corr12 = [d for d in data if d["corr"] == op_corr]
        if corr12:
            return corr12[0]["radi"]
        
        # Else take the highest corr value
        best = max(data, key=lambda x: x["corr"])
        return best["radi"]
    
    # Create new columns for each site
    for site in ["A1", "A2", "B1", "B2"]:
        elem_col = site
        ox_col = f"{site}_OS"
        new_col = f"S_IR-{site}"
        df[new_col] = df.apply(lambda row: get_radius(row[elem_col], row[ox_col]), axis=1)
    
    return df
```

```python
def filter(df_pred, lower_bandgap = 0.6, upper_bandgap = 3.0, octahedral_min=0.45, octahedral_max=0.9, tolerance_min=0.82, tolerance_max=1.08, new_tau_max=4.18):
    df = df_pred[
        (df_pred['PBE_bandgap'] >= lower_bandgap) & (df_pred['PBE_bandgap'] <= upper_bandgap) &
        (df_pred['octahedral_factor'] >= octahedral_min) & (df_pred['octahedral_factor'] <= octahedral_max) &
        (df_pred['tolerance_factor'] >= tolerance_min) & (df_pred['tolerance_factor'] <= tolerance_max) &
        (df_pred['stable'] == 1) & (df_pred['Insulator'] == 1) &
        (df_pred['new_tau'] < new_tau_max)
    ]
    return df
```
